```python
import numpy as np
import pandas as pd
from sklearn.model_selection import GridSearchCV, cross_val_predict, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import GradientBoostingClassifier
from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, log_loss, confusion_matrix
from sklearn.metrics import roc_auc_score
```

```python
models = {
    'Logistic Regression': {
        'model': LogisticRegression(max_iter=500),
        'param_grid': {
            'model__C': [0.1, 1, 10],
            'model__penalty': ['l2']
        }
    },
    'SVM': {
        'model': SVC(probability=True),
        'param_grid': {
            'model__C': [0.1, 1, 10],
            'model__kernel': ['linear', 'rbf'],
            'model__gamma': ['scale']
        }
    },
    'Gradient Boosted Trees': {
        'model': GradientBoostingClassifier(),
        'param_grid': {
            'model__n_estimators': [50, 100],
            'model__learning_rate': [0.01, 0.1],
            'model__max_depth': [3, 5]
        }
    },
    'XGBoost': {
        'model': XGBClassifier(use_label_encoder=False, eval_metric='mlogloss'),
        'param_grid': {
            'model__n_estimators': [50, 100],
            'model__learning_rate': [0.01, 0.1],
            'model__max_depth': [3, 5],
            'model__gamma': [0, 0.1]
        }
    }
}
```

```python
def nested_cv(X, y, models, outer_cv_folds=5, inner_cv_folds=5):
    results = {}
    
    outer_cv = StratifiedKFold(n_splits=outer_cv_folds, shuffle=True, random_state=42)
    
    for model_name, config in models.items():
        print(f"Processing {model_name}...")
        
        inner_cv = StratifiedKFold(n_splits=inner_cv_folds, shuffle=True, random_state=42)
        
        pipeline = Pipeline([
            ('scaler', StandardScaler()),
            ('model', config['model'])
        ])
        
        grid_search = GridSearchCV(
            estimator=pipeline,
            param_grid=config['param_grid'],
            scoring='accuracy',
            cv=inner_cv,
            n_jobs=-1
        )
        
        y_true, y_pred, y_pred_prob = [], [], []
        
        for train_idx, test_idx in outer_cv.split(X, y):
            X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
            y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
            
            grid_search.fit(X_train, y_train)
            best_model = grid_search.best_estimator_
            
            y_true.extend(y_test)
            y_pred.extend(best_model.predict(X_test))
            y_pred_prob.extend(best_model.predict_proba(X_test)[:, 1])
        
        y_true = np.array(y_true)
        y_pred = np.array(y_pred)
        y_pred_prob = np.array(y_pred_prob)
        
        metrics = {
            'mean_accuracy': accuracy_score(y_true, y_pred),
            'mean_precision': precision_score(y_true, y_pred),
            'mean_recall': recall_score(y_true, y_pred),
            'mean_f1_score': f1_score(y_true, y_pred),
            'mean_roc_auc': roc_auc_score(y_true, y_pred_prob),
            'mean_log_loss': log_loss(y_true, y_pred_prob)
        }
        
        results[model_name] = metrics
    
    return results
```

```python
X_train = X_train
y_train = train_data['Outcome']

cv_results = nested_cv(X_train, y_train, models)

for model_name, metrics in cv_results.items():
    print(f"{model_name}:")
    for metric_name, metric_value in metrics.items():
        print(f"  {metric_name}: {metric_value:.3f}")
```
